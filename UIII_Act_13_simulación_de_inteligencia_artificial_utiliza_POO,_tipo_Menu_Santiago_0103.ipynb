## 41. Credito autmotriz
#### Para simular la "Inteligencia Artificial" de una manera adecuada para preparatoria, añadiremos un método que actúa como un "Motor de Decisión Inteligente". Este motor evaluará el salario del alumno, el costo del auto y el enganche para predecir automáticamente si el crédito es Aprobado o Rechazado, simulando cómo los bancos usan algoritmos para tomar decisiones de riesgo.
![link text](https://raw.githubusercontent.com/SantiagoMedrano128/-UIII-Act-13-Proyecto-de-simulaci-n-de-inteligencia-artificial-utiliza-POO-tipo-Menu/refs/heads/main/infoact12santiago1013.png)

```python
# ==============================
# SIMULADOR IA CRÉDITO AUTOMOTRIZ
# Programación Orientada a Objetos
# ==============================

class CreditoAutomotriz:

    # Función constructora
    def __init__(self, salariototal, preciovehiculo,
                 enganche, plazomeses, modeloymarca):

        # Atributos solicitados
        self.salariototal = float(salariototal)
        self.preciovehiculo = float(preciovehiculo)
        self.enganche = float(enganche)
        self.plazomeses = int(plazomeses)
        self.modeloymarca = modeloymarca

        # Atributos adicionales
        self.montoFinanciado = self.calcularMontoFinanciado()
        self.cuotaMensual = self.calcularCuotaMensual()
        self.saldoPendiente = self.montoFinanciado


    # Función 1
    def calcularMontoFinanciado(self):

        return self.preciovehiculo - self.enganche


    # Función 2
    def calcularCuotaMensual(self):

        monto = self.calcularMontoFinanciado()

        # Interés simple anual del 12%
        interes = monto * 0.12

        totalPagar = monto + interes

        return totalPagar / self.plazomeses


    # Función 3
    def validarEngancheMinimo(self):

        engancheMinimo = self.preciovehiculo * 0.20

        return self.enganche >= engancheMinimo


    # Función 4
    def registrarPago(self, pago):

        pago = float(pago)

        if pago <= self.saldoPendiente:

            self.saldoPendiente = self.saldoPendiente - pago

            print("✅ Pago registrado correctamente")

            print("Saldo restante: $",round(self.saldoPendiente,2))

        else:

            print("❌ El pago excede el saldo")


    # Función 5
    def iaMotorDecision(self):

        if self.validarEngancheMinimo()==False:

            return "Rechazado: Enganche insuficiente"

        elif self.cuotaMensual > self.salariototal*0.35:

            return "Rechazado: Riesgo alto"

        else:

            return "Aprobado"


# ==============================
# LISTA DE CRÉDITOS
# ==============================

listaCreditos=[]


# ==============================
# TABLA
# ==============================

def mostrarTabla():

    print("\n")
    print("="*105)

    print(f"{'ID':<5}{'AUTO':<20}{'PRECIO':<15}{'ENGANCHE':<15}{'MENSUALIDAD':<18}{'IA':<25}")

    print("="*105)

    for i,credito in enumerate(listaCreditos):

        print(f"{i:<5}"
              f"{credito.modeloymarca:<20}"
              f"${credito.preciovehiculo:<14,.0f}"
              f"${credito.enganche:<14,.0f}"
              f"${credito.cuotaMensual:<17,.2f}"
              f"{credito.iaMotorDecision():<25}")

    print("="*105)



# ==============================
# MENÚ PRINCIPAL
# ==============================

while True:

    print("\n")
    print("🤖==============================🤖")
    print(" SISTEMA IA CRÉDITO AUTOMOTRIZ ")
    print("🤖==============================🤖")

    print("1. Registrar crédito")
    print("2. Ver créditos")
    print("3. Registrar pago")
    print("4. Salir")

    opcion=input("Selecciona opción: ")


    # Opción 1

    if opcion=="1":

        print("\n--- NUEVO CRÉDITO ---")

        salario=input("Salario mensual: ")
        precio=input("Precio del vehículo: ")
        enganche=input("Enganche: ")
        plazo=input("Plazo meses: ")
        modelo=input("Marca y modelo: ")

        if salario.replace(".","").isdigit() and precio.replace(".","").isdigit() and enganche.replace(".","").isdigit() and plazo.isdigit():

            nuevo=CreditoAutomotriz(
                salario,
                precio,
                enganche,
                plazo,
                modelo
            )

            listaCreditos.append(nuevo)

            print("\n🧠 Analizando con IA...")

            print("Resultado:",
                  nuevo.iaMotorDecision())

        else:

            print("❌ Solo números válidos")


    # Opción 2

    elif opcion=="2":

        if len(listaCreditos)==0:

            print("No existen créditos")

        else:

            mostrarTabla()


    # Opción 3

    elif opcion=="3":

        if len(listaCreditos)==0:

            print("No existen créditos")

        else:

            mostrarTabla()

            idCredito=input("ID del crédito: ")

            if idCredito.isdigit():

                idCredito=int(idCredito)

                if idCredito>=0 and idCredito<len(listaCreditos):

                    pago=input("Cantidad a pagar: ")

                    if pago.replace(".","").isdigit():

                        listaCreditos[idCredito].registrarPago(pago)

                    else:

                        print("❌ Pago inválido")

                else:

                    print("❌ ID inexistente")

            else:

                print("❌ Solo números")


    # Opción 4

    elif opcion=="4":

        print("\n🚗 Gracias por usar el sistema IA")
        break


    else:

        print("❌ Opción incorrecta")
```
